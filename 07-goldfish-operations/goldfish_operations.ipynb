```python
import numpy as np
from PIL import Image
import cv2 as cv2
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import os.path as osp
```

```python
goldfish_bgr = cv2.imread(osp.join('images', 'goldfish.jpg'))
goldfish_rgb = goldfish_bgr[:,:,::-1] #Reverse order in the third dimension to convert from BGR to RGB.
plt.figure()
plt.imshow(goldfish_rgb)
```

```python
inverted_image = 255 - goldfish_rgb #Compute the missing values, subtract 255 out of every value of the colour channel.
plt.figure()
plt.imshow(inverted_image)
```

```python
flipped_goldfish = goldfish_rgb[:, ::-1, :]
plt.figure()
plt.imshow(flipped_goldfish) #Flipping an image is simply flipping the columns.
```

```python
goldfish_grayscale = cv2.cvtColor(goldfish_rgb, cv2.COLOR_RGB2GRAY)
plt.figure()
plt.imshow(goldfish_grayscale, cmap='gray', vmin=0, vmax=255)
```

```python
addition_goldfish = goldfish_grayscale + 30
plt.figure()
plt.imshow(addition_goldfish, cmap='gray', vmin=0, vmax=255)
```

```python
goldfish_red_channel = goldfish_rgb.copy()
goldfish_red_channel[:,:,[1,2]] = 0
plt.figure()
plt.imshow(goldfish_red_channel)
```

```python
#Splitting an image
midpoint = goldfish_rgb.shape[1] // 2
left_goldfish = goldfish_rgb[:,:midpoint,:]
right_goldfish = goldfish_rgb[:,midpoint:,:]

plt.figure()
plt.imshow(left_goldfish)
```

```python
plt.figure()
plt.imshow(right_goldfish)
```

```python
top_crop = goldfish_rgb.copy()
top_crop[:100, :, :] = 0 #Remove the first 100 rows.
plt.figure()
plt.imshow(top_crop)
```
