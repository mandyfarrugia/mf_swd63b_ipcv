```python
import numpy as np
from math import gcd

#Aspect ratio - the relationship between width and height of an image, screen, or object

#A function that returns a ratio in decimal format.
def calculate_aspect_ratio_decimals(width, height):
    div = width / height
    return f'1:{div}'

#A function that returns a ratio of one value to another.
def calculate_aspect_ratio(width, height):
    divisor = gcd(width, height) #Computing the common denominator/divisor between the width and height.
    first_number = width / divisor
    second_number = height / divisor
    return f'{first_number}:{second_number}'
```

```python
calculate_aspect_ratio_decimals(1920, 1080)
```

```python
calculate_aspect_ratio(1920, 1080)
```
