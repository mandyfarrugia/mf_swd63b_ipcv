```python
import numpy as np
from PIL import Image
import cv2 as cv2
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import os.path as osp
```

```python
path_to_image = osp.join('images', 'lena.jpg')
image_bgr = cv2.imread(path_to_image)
plt.figure()
plt.imshow(image_bgr)
```

```python
image_rgb=image_bgr[:,:,::-1] #Swap the colour channels using numpy channel slicing.
#or use cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
plt.figure()
plt.imshow(image_rgb)
```

```python
image_grayscale = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)
plt.figure()
plt.imshow(image_grayscale, cmap='gray', vmin=0, vmax=255) #Use 'gray' instead of 'Grays'.
```

```python
image_in_red = image_rgb.copy()
plt.figure()
image_in_red[:,:,[1,2]] = 0 #Set the blue and green components to 0 to view only the red channels of the image.
plt.imshow(image_in_red)
```
