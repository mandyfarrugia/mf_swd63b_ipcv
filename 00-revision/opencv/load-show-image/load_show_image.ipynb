# How to load an show an image using OpenCV and Python?

## Library Imports

Start by importing the OpenCV library, which is an image processing library, then give it the alias ```cv```.

````python
import cv2 as cv
import os.path as osp
````

## Load an image using OpenCV

````python
path_to_image_to_show = osp.join('images', 'crushed_can.jpg')
loaded_image = cv.imread(path_to_image_to_show) #Load image using the imread function within the OpenCV library.
loaded_image
````

## Load a preview of the image using OpenCV

````python
cv.imshow('Image of crushed can', loaded_image)

'''
The preview window must be closed explicitly otherwise a crash will be encountered
when executing this code from an interactive shell such as a Jupyter notebook.
This crash happens because the preview window uses the same Python process as the kernel,
hence Jupyter will remain waiting indefinitely, causing a crash.
'''

cv.waitKey(0) #A parameter of 0 waits for any key to be pressed.
cv.destroyAllWindows() #Explicitly instruct Python to close all instances of preview windows.

#Reference: https://medium.com/@mrdatainsight/how-to-use-opencv-imshow-in-a-jupyter-notebook-quick-tip-ce83fa32b5ad
````
