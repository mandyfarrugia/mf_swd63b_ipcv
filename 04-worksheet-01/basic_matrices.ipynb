```python
width_pixels = 100
height_pixels = 200

matrix_M3 = np.zeros([width_pixels, height_pixels, 3], dtype='uint8')

orange_rgb_values = [255, 128, 0]
blue_rgb_values = [0, 0, 255]

matrix_M3[:, :100] = orange_rgb_values #For each row, ensure that columns starting from location 0 up to 99 are in orange.
matrix_M3[:, 100:] = blue_rgb_values #For each row, ensure that columns starting from location 100 up to 199 are in blue.
```

```python
plt.imshow(matrix_M3, vmin=0, vmax=255)
```

```python
Image.fromarray(matrix_M3).save('half_orange_half_blue.png')
```
