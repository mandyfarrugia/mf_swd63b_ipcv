```python
import numpy as np
from PIL import Image
import cv2 as cv2
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import os.path as osp
```

# Task 1

```python
image_bgr = cv2.imread(osp.join('images', 'lena.jpg'))
image_grayscale = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)
grayscale_matrix = np.array(image_grayscale)
plt.imshow(grayscale_matrix, cmap='gray', vmin=0, vmax=255)
```

```python
try:
    t_value = int(input('Enter a value: '))

    if(t_value < 0 or t_value > 255):
        raise Exception('Values less than 0 or larger than 255 are not accepted!')

    image_inversion_result = t_value - grayscale_matrix
    plt.imshow(image_inversion_result)
except ValueError:
    print('Non-numeric input is not accepted!')
except Exception as base_exception:
    print(base_exception)
```

# Task 2

```python
goldfish_image_rgb = Image.open(osp.join('images', 'goldfish.jpg')).convert('RGB')
goldfish_image_rgb_matrix = np.array(goldfish_image_rgb)
plt.imshow(goldfish_image_rgb_matrix)
```

```python
red_channel_matrix = goldfish_image_rgb_matrix.copy()
red_channel_matrix[:,:,[1,2]] = 0
plt.imshow(red_channel_matrix)
```

```python
green_channel_matrix = goldfish_image_rgb_matrix.copy()
green_channel_matrix[:,:,[0,2]] = 0
plt.imshow(green_channel_matrix)
```

```python
blue_channel_matrix = goldfish_image_rgb_matrix.copy()
blue_channel_matrix[:,:,[0,1]] = 0
plt.imshow(blue_channel_matrix)
```

```python
plt.figure()

plt.subplot(2,2,1)
plt.imshow(goldfish_image_rgb_matrix)
plt.title('Original Image')
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(red_channel_matrix)
plt.title('Red Channel Matrix')
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(green_channel_matrix)
plt.title('Green Channel Matrix')
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(blue_channel_matrix)
plt.title('Blue Channel Matrix')
plt.axis('off')

plt.show()
```

```python
Image.fromarray(red_channel_matrix).save(osp.join('images', 'red_channel_goldfish.png'))
```

```python
Image.fromarray(green_channel_matrix).save(osp.join('images', 'green_channel_goldfish.png'))
```

```python
Image.fromarray(blue_channel_matrix).save(osp.join('images', 'blue_channel_goldfish.png'))
```

# Task 3

```python
lena_image_rgb = Image.open(osp.join('images', 'lena.jpg')).convert('RGB')
lena_image_rgb_matrix = np.array(lena_image_rgb)
```

```python
width = lena_image_rgb_matrix.shape[0]
height = lena_image_rgb_matrix.shape[1]
colour_channel = lena_image_rgb_matrix.shape[2]
(width, height, colour_channel)
```

```python
midpoint_width = width // 2
midpoint_height = height // 2
```

```python
top_left = lena_image_rgb_matrix[0: midpoint_height, 0: midpoint_width]
top_right = lena_image_rgb_matrix[0: midpoint_height, midpoint_width: width]
bottom_left = lena_image_rgb_matrix[midpoint_height: height, 0: midpoint_width]
bottom_right = lena_image_rgb_matrix[midpoint_height: height, midpoint_width: width]
```

```python
plt.figure()

plt.subplot(2, 2, 1)
plt.imshow(top_left)
plt.title("Top Left")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(top_right)
plt.title("Top Right")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(bottom_left)
plt.title("Bottom Left")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(bottom_right)
plt.title("Bottom Right")
plt.axis("off")

plt.show()
```

```python
quad_border_crop = lena_image_rgb_matrix.copy()
quad_border_crop[:100, :, :] = 0
quad_border_crop[-100:, :, :] = 0
quad_border_crop[:, :100, :] = 0
quad_border_crop[:, -100:, :] = 0

plt.figure()
plt.imshow(quad_border_crop)
plt.axis("off")
plt.show()
```

```python
Image.fromarray(quad_border_crop).save(osp.join('images', 'quad_crop.png'))
```
