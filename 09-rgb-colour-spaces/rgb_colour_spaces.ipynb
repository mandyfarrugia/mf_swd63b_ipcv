```python
import numpy as np
from PIL import Image
import cv2 as cv2
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import os.path as osp
```

```python
def my_nothing(x):
    pass
```

```python
colour_chart = np.zeros((300, 512, 3), np.uint8)
cv2.namedWindow('Colour_window')
cv2.createTrackbar('R', 'Colour_window', 0, 255, my_nothing)
cv2.createTrackbar('G', 'Colour_window', 0, 255, my_nothing)
cv2.createTrackbar('B', 'Colour_window', 0, 255, my_nothing)

switch = '0 : OFF \n 1 : ON'
cv2.createTrackbar(switch, 'Colour_window', 0, 1, my_nothing)

while(1):
    cv2.imshow('Colour_window', colour_chart) # cv2.imshow accepts BGR as default.
    k = cv2.waitKey(1) & 0xFF

    if k == 27:
        break

    r_value = cv2.getTrackbarPos('R', 'Colour_window')
    g_value = cv2.getTrackbarPos('G', 'Colour_window')
    b_value = cv2.getTrackbarPos('B', 'Colour_window')
    s_value = cv2.getTrackbarPos(switch, 'Colour_window')

    if(s_value == 0):
        colour_chart[:] = 0
    else:
        colour_chart[:] = [b_value, g_value, r_value]

cv2.destroyAllWindows()
```
